#### Question 1
#### Employee Data Bank.



Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:
    def __init__(self,name,number,address,salary,post,year_of_joining,education,year_of_experience):
        self.name=name
        self.number=number
        self.address=address
        self.salary=salary
        self.post=post
        self.year_of_joining=year_of_joining
        self.education=education
        self.year_of_experience=year_of_experience
    def describe_employee(self):
        print(f'Name = {self.name}')
        print(f'Number = {self.number}')
        print(f'Address = {self.address}')
        print(f'Salary = {self.salary}')
        print(f'Post = {self.post}')
        print(f'Year of Joining = {self.year_of_joining}')
        print(f'Education = {self.education}')
        print(f'Year of Experience = {self.year_of_experience} Years')
    def update_year_of_experience(self):
        self.year_of_experience += 1
        print(f'Update Year of Experience = {self.year_of_experience}')
    def increase_salary(self,amount):
        self.salary += amount
        print(f'Updated Salary = {self.salary}')
```

```python
#INSTANCE

employee_101 = Employee("Isha Naz",29154,"Karachi",74000,"Medical Technologist",2022,"BS",5) 
```

```python
#printing four attri_butes

print(employee_101.name)
print(employee_101.number)
print(employee_101.salary)
print(employee_101.education)
```

```python
#Calling methods

employee_101.describe_employee()
employee_101.update_year_of_experience()  #Updating year of experience
employee_101.increase_salary(16000)   #salary after increment
```

```python
#Creating 10 instances 

employee_01 = Employee("Ahmed",111,"Karachi",40000,"Receptionist",2023,"BA",3)
employee_02 = Employee("Ali",112,"Lahore",20000,"House keeper",2023,"Matric",4)
employee_03 = Employee("Arsalan",113,"Karachi",25000,"Security guard",2023,"Matric",4)
employee_04 = Employee("Uzair",114,"Peshawar",55000,"Inventory Officer",2023,"BA",4)
employee_05 = Employee("Fahad",115,"Karachi",25000,"Security guard",2023,"Matric",4)
employee_06 = Employee("Taimoor",116,"Hyderabad",25000,"Security guard",2023,"Matric",4)
employee_07 = Employee("Ehsan",117,"Karachi",20000,"House keeper",2023,"Matric",4)
employee_08 = Employee("Nameer",118,"Faisalabad",20000,"House keeper",2023,"Matric",4)
employee_09 = Employee("Faisal",119,"Islamabad",95000,"Supervisor",2025,"MBA",1)
employee_10 = Employee("Fahad",120,"Karachi",30000,"Receptionist",2023,"BA",4)
```

```python
# Put all 10 employees into a dictionary so we can loop over them

employees = {
    "employee_01": employee_01,
    "employee_02": employee_02,
    "employee_03": employee_03,
    "employee_04": employee_04,
    "employee_05": employee_05,
    "employee_06": employee_06,
    "employee_07": employee_07,
    "employee_08": employee_08,
    "employee_09": employee_09,
    "employee_10": employee_10,
}
```

```python
#Calling each employee by using describe_employee

employee_01.describe_employee()
print()
employee_02.describe_employee()
print()
employee_03.describe_employee()
print()
employee_04.describe_employee()
print()
employee_05.describe_employee()
print()
employee_06.describe_employee()
print()
employee_07.describe_employee()
print()
employee_08.describe_employee()
print()
employee_09.describe_employee()
print()
employee_10.describe_employee()
print()
```

```python
#Applying methods on all employee by using for loop

for employee in employees.values():
    employee.describe_employee()
    employee.update_year_of_experience()
    employee.increase_salary(2500)
    print()
```

#### Question 2

Complete the same task (Question 1) without using the Classes or OOP.

```python
def describe_employee(emp):
    print(f"Name = {emp['name']}")
    print(f"Number = {emp['number']}")
    print(f"Address = {emp['address']}")
    print(f"Salary = {emp['salary']}")
    print(f"Post = {emp['post']}")
    print(f"Year of Joining = {emp['year_of_joining']}")
    print(f"Education = {emp['education']}")
    print(f"Year of Experience = {emp['year_of_experience']} Years")


def update_year_of_experience(emp):
    emp["year_of_experience"] += 1
    print(f"Updated Experience = {emp['year_of_experience']} Years")


def increase_salary(emp, amount):
    emp["salary"] += amount
    print(f"Updated Salary = {emp['salary']}")
```

```python
# INSTANCE

employee_101 = {
    "name": "ISHA NAZ",
    "number": 29154,
    "address": "Karachi",
    "salary": 74000,
    "post": "Medical Technologist",
    "year_of_joining": 2022,
    "education": "BS",
    "year_of_experience": 5}
```

```python
#printing four attri_butes

print(employee_101["name"])
print(employee_101["number"])
print(employee_101["address"])
print(employee_101["salary"])
```

```python
#calling Functions

describe_employee(employee_101)
update_year_of_experience(employee_101)
increase_salary(employee_101,16000)
```
