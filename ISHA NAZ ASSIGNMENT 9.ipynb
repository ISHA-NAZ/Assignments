```python
import csv

with open("Student Data.csv", "r") as file:
    reader = csv.DictReader(file)
    students = list(reader)

for student in students:
    Percentage = float(student["Percentage"])

    if Percentage >= 80:
        student["Grade"] = "A"
    elif Percentage >= 70:
        student["Grade"] = "B"
    elif Percentage >= 60:
        student["Grade"] = "C"
    elif Percentage >= 50:
        student["Grade"] = "D"
    else:
        student["Grade"] = "F"

with open("student_data.csv", "w", newline="") as file:
    fieldnames = students[0].keys()
    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(students)

print("Grades added successfully!")
```
