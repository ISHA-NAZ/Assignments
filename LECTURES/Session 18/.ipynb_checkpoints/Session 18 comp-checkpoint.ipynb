### **Retrieve all rows where there is a matching record in both the facts and cities tables, using the id column in the facts table and the facts_id column in the cities table as the join condition.**

```python
q = ''' select * from facts 
        inner join cities
        on 
        facts.id = cities.facts_id


'''
sql_query(q)
```

```python
q = ''' select * from facts f
        inner join cities c
        on 
        f.id = c.facts_id


'''
sql_query(q)
```

### **Show the country name and city name for entries where the country is 'Pakistan'**

```python
q = """
    select f.name as Country_name, c.name as cities_name from facts f
    inner join cities c
    on
    f.id = c.facts_id
    where f.name = 'Pakistan'
    """
sql_query(q)
```

### **find countries that do not have any cities listed in the cities table. The query should return the country name and city name.**

```python
q = """
    select f.name as Country_name, c.name as cities_name from facts f
    left join cities c
    on
    f.id = c.facts_id
    where c.name is null
    """
sql_query(q)
```

```python
q = """
    select f.name as Country_name, c.name as cities_name from cities c
    right join facts f
    on
    f.id = c.facts_id
    where c.name is null
    """
sql_query(q)
```

### **Show Countryname cityname and citypopulation contains capital == 1, citypopulation > 10000000 ten million**

```python
q = """ select f.name Country_Name, c.name City_Name, c.population City_population from facts f
        inner join cities c
        on 
        f.id = c.facts_id
        where c.capital == 1 and c.population > 10000000


"""
sql_query(q)
```

### **Show the country name and city name for entries where the country is 'United Arab Emirates'**


```python
q = ''' select f.name Country_name, c.name City_name from facts f
        inner join cities c
        on
        f.id = c.facts_id
        where f.name = 'United Arab Emirates' 

'''
sql_query(q)
```

###  **Show Countryname Top 3 cities population Where capital == 0 and population > 10000000**

```python
q = ''' select f.name country_name, c.name city_name, c.population city_population from facts f
        inner join cities c
        on
        f.id = c.facts_id
        where c.capital == 1 and c.population > 10000000
        order by 3 desc limit 3

'''
sql_query(q)
```
